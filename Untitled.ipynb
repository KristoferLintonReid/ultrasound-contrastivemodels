```python
import os
import pandas as pd
import torch
import nibabel as nib
import numpy as np
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from sklearn.model_selection import train_test_split
import torch.nn as nn
```

```python
# Step 1: Load the RNA-Seq data (ENS gene IDs only)
df_rna = pd.read_csv('RNA_counts_141024.csv', header=None, low_memory=False)

# Step 2: Transpose and clean up the data
df_rna_transposed = df_rna.transpose()

# Make the second row the header and remove unwanted rows
new_header = df_rna_transposed.iloc[1]  # First row for the header
df_rna_transposed = df_rna_transposed[2:]  # Data without header rows
df_rna_transposed.columns = new_header  # Set the header row

# Step 3: Filter columns that contain 'ENSG' substring (gene IDs)
rna_cols = df_rna_transposed.filter(like='ENSG')  # Only columns with 'ENSG' in their name

# Step 4: Load and clean the image data linking file
df_images = pd.read_excel('RNA_Sample_Linking2.xlsx')
df_images['Sample ID'] = df_images['Sample ID'].str.replace(' ', '_')

# Step 5: Merge the RNA data with the image data
df_merged = pd.merge(df_images, df_rna_transposed, left_on='Sample ID', right_on='Geneid')

# Directory for images
image_directory = '/home/kl2418/Documents/Barcroft/CLIPRNA/images_matched/'

# Step 6: Prepare dataset and dataloaders
# Convert 'CASE ID' and 'CASE ID2' to strings and safely handle missing or NaN values
image_filenames = []
rna_data_filtered = []

# Limit to 20 pairs
max_pairs = 20
pair_count = 0

for index, row in df_merged.iterrows():
    if pair_count >= max_pairs:
        break  # Stop after collecting 20 pairs
    
    case_id = row['CASE ID ']
    case_id2 = row['CASE ID2']
    
    # Convert to string and strip any whitespace, handle NaN values
    case_id_str = str(case_id).strip() if pd.notna(case_id) else ''
    case_id2_str = str(case_id2).strip() if pd.notna(case_id2) else ''
    
    # Check if a substring of case_id or case_id2 is in any image filename, exclude files with 'seg'
    for file_name in os.listdir(image_directory):
        if file_name.endswith('.nii.gz') and 'seg' not in file_name:  # Exclude 'seg' files (binary masks)
            if case_id_str in file_name or case_id2_str in file_name:
                image_filenames.append(file_name)
                
                # Append the corresponding RNA data (filtered for 'ENSG' columns)
                rna_data_filtered.append(row[rna_cols.columns].apply(pd.to_numeric, errors='coerce').fillna(0))
                
                pair_count += 1
                break  # Stop searching further after a match for this row

# Debug: Print number of image files found
print(f"Number of matching image files found (excluding 'seg' and using max of 20): {len(image_filenames)}")
print(f"Number of RNA samples corresponding to images: {len(rna_data_filtered)}")

# Convert RNA data to a DataFrame for easy handling
rna_data_filtered = pd.DataFrame(rna_data_filtered)

# Check if we have 20 pairs
if len(rna_data_filtered) == 0:
    raise ValueError("No matching RNA data found for the corresponding images.")
if len(image_filenames) != len(rna_data_filtered):
    raise ValueError("Mismatch between the number of RNA samples and image files.")
```

```python
# Step 9: Proceed with splitting the data for training
# Split into train and test (with the limited subset of 20 pairs)
rna_train, rna_test, img_train, img_test = train_test_split(rna_data_filtered, image_filenames, test_size=0.2)

# Step 10: Prepare datasets and dataloaders
train_dataset = RNACustomDataset(rna_train, img_train, transform=transform)
test_dataset = RNACustomDataset(rna_test, img_test, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True)  # Adjust batch size
test_loader = DataLoader(test_dataset, batch_size=4, shuffle=False)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader

# Step 1: Define the RNA Encoder (Tabular model)
class RNAEncoder(nn.Module):
    def __init__(self, input_dim, embedding_dim=512):
        super(RNAEncoder, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(input_dim, 1024),
            nn.ReLU(),
            nn.Linear(1024, 512),
            nn.ReLU(),
            nn.Linear(512, embedding_dim)
        )
        
    def forward(self, x):
        return self.model(x)

# Step 2: Define the Image Encoder (CNN model)
from torchvision.models import ResNet50_Weights
from torchvision import models

class ImageEncoder(nn.Module):
    def __init__(self, embedding_dim=512):
        super(ImageEncoder, self).__init__()
        # Use the updated weights parameter to avoid the deprecation warning
        resnet = models.resnet50(weights=ResNet50_Weights.DEFAULT)
        resnet.fc = nn.Linear(resnet.fc.in_features, embedding_dim)  # Adjust final layer to match embedding_dim
        self.model = resnet
    
    def forward(self, x):
        return self.model(x)

# Step 3: Define the Combined CLIP-like Model
class CLIPModel(nn.Module):
    def __init__(self, rna_encoder, image_encoder):
        super(CLIPModel, self).__init__()
        self.rna_encoder = rna_encoder
        self.image_encoder = image_encoder
    
    def forward(self, rna_input, image_input):
        rna_embeddings = self.rna_encoder(rna_input)
        image_embeddings = self.image_encoder(image_input)
        return rna_embeddings, image_embeddings

# Step 4: Contrastive Loss (NT-Xent)
class ContrastiveLoss(nn.Module):
    def __init__(self, temperature=0.5):
        super(ContrastiveLoss, self).__init__()
        self.temperature = temperature
        self.cosine_similarity = nn.CosineSimilarity(dim=-1)
    
    def forward(self, rna_embeddings, image_embeddings):
        # Normalize embeddings
        rna_embeddings = nn.functional.normalize(rna_embeddings, p=2, dim=-1)
        image_embeddings = nn.functional.normalize(image_embeddings, p=2, dim=-1)
        
        # Compute similarity matrix
        similarity_matrix = self.cosine_similarity(rna_embeddings.unsqueeze(1), image_embeddings.unsqueeze(0)) / self.temperature
        
        # Contrastive loss (NT-Xent)
        labels = torch.arange(rna_embeddings.size(0)).to(rna_embeddings.device)
        loss_fn = nn.CrossEntropyLoss()
        loss = (loss_fn(similarity_matrix, labels) + loss_fn(similarity_matrix.t(), labels)) / 2
        
        return loss

# Step 5: Training the Combined Model with Validation
def train(model, train_loader, val_loader, optimizer, criterion, device, epochs):
    for epoch in range(epochs):
        # Training Phase
        model.train()  # Set both encoders to training mode
        total_train_loss = 0
        for rna_batch, image_batch in train_loader:
            rna_batch, image_batch = rna_batch.to(device), image_batch.to(device)
            
            # Forward pass through both encoders
            rna_embeddings, image_embeddings = model(rna_batch, image_batch)
            
            # Compute contrastive loss
            loss = criterion(rna_embeddings, image_embeddings)
            total_train_loss += loss.item()
            
            # Backward pass and optimization
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
        
        avg_train_loss = total_train_loss / len(train_loader)
        
        # Validation Phase (no gradient calculation)
        model.eval()
        total_val_loss = 0
        with torch.no_grad():  # Disable gradient calculation
            for rna_batch, image_batch in val_loader:
                rna_batch, image_batch = rna_batch.to(device), image_batch.to(device)
                
                # Forward pass through both encoders
                rna_embeddings, image_embeddings = model(rna_batch, image_batch)
                
                # Compute contrastive loss
                loss = criterion(rna_embeddings, image_embeddings)
                total_val_loss += loss.item()
        
        avg_val_loss = total_val_loss / len(val_loader)
        
        # Print both training and validation loss for each epoch
        print(f'Epoch {epoch+1}/{epochs}, Training Loss: {avg_train_loss:.4f}, Validation Loss: {avg_val_loss:.4f}')
```

```python
# Step 6: Initialize the Model, Optimizer, and Criterion
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Initialize the encoders
rna_encoder = RNAEncoder(input_dim=len(rna_cols.columns), embedding_dim=512)
image_encoder = ImageEncoder(embedding_dim=512)

# Combine the encoders into a single model
model = CLIPModel(rna_encoder, image_encoder).to(device)

# Define optimizer and loss function
optimizer = optim.Adam(model.parameters(), lr=1e-4)
criterion = ContrastiveLoss(temperature=0.5)

# Step 7: Train the Model
epochs = 10
train(model, train_loader, test_loader, optimizer, criterion, device, epochs)

```
