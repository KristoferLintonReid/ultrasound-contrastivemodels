```python
import pandas as pd
import os
import shutil

# Step 1: Read the CSV file into a DataFrame without setting the first row as the header
# Use low_memory=False to avoid DtypeWarning
df = pd.read_csv('RNA_counts_141024.csv', header=None, low_memory=False)

# Step 2: Transpose the DataFrame
df_transposed = df.transpose()

# Step 3: Make the first row the header
new_header = df_transposed.iloc[1]  # First row for the header
df_transposed = df_transposed[2:]    # Take the data less the header row
df_transposed.columns = new_header  # Set the header row as the dataframe header

# Read the Excel file and adjust 'Sample ID' for merging
df2 = pd.read_excel('RNA_Sample_Linking2.xlsx')
df2['Sample ID'] = df2['Sample ID'].str.replace(' ', '_')

# Merge DataFrames
df3 = pd.merge(df2, df_transposed, left_on='Sample ID', right_on='Geneid')

# Define directories
source_directory = '/home/kl2418/Documents/Barcroft/Barcroft_Data_FINAL/'
destination_directory = '/home/kl2418/Documents/Barcroft/CLIPRNA/images_matched/'

# Ensure destination directory exists
if not os.path.exists(destination_directory):
    os.makedirs(destination_directory)

# Iterate over rows in df3 and match CASE ID or CASE ID2 to image files
for index, row in df3.iterrows():
    case_id = str(row['CASE ID ']) if not pd.isna(row['CASE ID ']) else ''  # Ensure case_id is a string
    case_id2 = str(row['CASE ID2']) if not pd.isna(row['CASE ID2']) else ''  # Ensure case_id2 is a string
    
    # Find matching .nii.gz files in the source directory
    for file_name in os.listdir(source_directory):
        if file_name.endswith('.nii.gz'):
            if case_id in file_name or case_id2 in file_name:
                # Full path of the source and destination
                source_file_path = os.path.join(source_directory, file_name)
                destination_file_path = os.path.join(destination_directory, file_name)
                
                # Copy the file to the destination directory
                shutil.copy(source_file_path, destination_file_path)
                print(f"Copied {file_name} to {destination_directory}")
```
